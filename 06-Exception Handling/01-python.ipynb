## Exception handling example

```python
try:
    a=b
except:
    print("Error agaya bhai. Variable not declared")
```

```python
try:
    a=b  # NameError will be raised
except NameError as e:
    print("Caught an exception:", e)
```

```python
try:
    print(1/0)  # ZeroDivisionError
except ZeroDivisionError as E:
    print("Error agaya: ",E)
```

```python
try:
    r=1/0
    t=r
# except ZeroDivisionError as E:
#     print(E)
except Exception as E:              # Main Exception, pos=" ! last "
    print(E)
```

```python
try:
    r=int(input("Enter Number only, num > 0"))
    t=1/r
except ZeroDivisionError as E:
    print("Error:",E)
except ValueError:
    print("Error: ",r," is not a num")
except Exception as E:              # Main Exception, pos=" ! last "
    print("Error:",E)
else:
    print(f"Value is: {t}")
finally:
    print("Opr completed")
```
