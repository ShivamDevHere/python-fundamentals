```python
def Greet(name="Guest"):
    print(f"Welcome {name} to us")
Greet()

def Collection(*args):                      #Positional arguments
    [print(number, end="  ") for number in args]
Collection(1,2,3,4,5,5,5,5,2,323,32,3,'YOOOOO')
print()

def details(**kwargs):              # Keyword Arguments
    for key,value in kwargs.items():
        print(f"{key}:{value}")
details(name="Guest", Age=22, State="Learning AI")
```

```python
def details(*args,**kwargs):      
    [print(number, end="  ") for number in args]
    print()
    for key,value in kwargs.items():
        print(f"{key}:{value}")
details(1,2,3,4,5,5,5,5,2,323,32,3,'YOOOOO',name="Guest", Age=22, State="Learning AI")
```
