```python
numbers = [1,2,3]
text = ['apple','hello0','gross']
print(list(map(lambda x:x**2,numbers)))
print(list(map(lambda x:x**2,[1,2,3])))
print(list(map(str.upper,text)))
```

```python
def list_dict(dict):
    return dict['name']
dict = [
    {'name':'Jack','age':34},
    {'name':'Alexa','age':32}
]
print(list(map(list_dict,dict)))
```

```python
# Map <----iterator
# Filter <-iterator + returns only True

print(list(filter(lambda x:x%2==0 and x>7,range(0,20))))

def list_dict(dict):
    return dict['age']>18
dict = [
    {'name':'Jack','age':8},
    {'name':'Alexa','age':32},
    {'name':'Widodni','age':16},
    {'name':'carryat','age':78}
]
print(list(filter(list_dict,dict)))
```
