# LISTS
---

``` python
arr = [10, 20, 30]
```

### Adding Operations

-   End Insertion: `arr.append(50)`
-   Add Position: `arr.insert(3, 40)`
-   Add Multiple: `arr.extend([60, 70])`

### Removing Operations

-   Remove Value: `arr.remove(20)`
-   Remove Last: `arr.pop()`
-   Remove Index: `arr.pop(1)`
-   Delete: `del arr[0]`
-   Clear: `arr.clear()`

### Ordering Operations

-   Sort: `arr.sort()`
-   Reverse: `arr.reverse()`

### Searching Operations

-   Count: `arr.count(20)`
-   Find Index: `arr.index(30)`
-   Check: `20 in arr`

### Other Operations

-   Copy: `arr.copy()`
-   Length: `len(arr)`
-   Slice: `arr[1:3]`

------------------------------------------------------------------------
