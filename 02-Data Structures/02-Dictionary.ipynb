# 2. Dictionary

``` python
student = {"name": "Shivam", "age": 23}
```

### Adding / Updating Operations

-   Add Key: `student["city"] = "Delhi"`
-   Update Value: `student["age"] = 22`
-   Add Multiple: `student.update({"city": "Delhi", "age": 22})`

### Removing Operations

-   Remove Key: `student.pop("age")`
-   Remove Last: `student.popitem()`
-   Delete: `del student["name"]`
-   Clear: `student.clear()`

### Accessing Operations

-   Access Value: `student["name"]`
-   Safe Access: `student.get("name")`
-   Get Keys: `student.keys()`
-   Get Values: `student.values()`
-   Get Items: `student.items()`

### Other Operations

-   Copy: `student.copy()`
-   Check Key: `"name" in student`
-   Length: `len(student)`

------------------------------------------------------------------------
