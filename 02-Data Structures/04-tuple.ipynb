# Tuple

``` python
t = (10, 20, 30, 40)
```

### Accessing Operations

-   Access: `t[0]`
-   Check: `20 in t`
-   Slice: `t[1:3]`
-   Length: `len(t)`

### Searching Operations

-   Count: `t.count(20)`
-   Find Index: `t.index(30)`

### Conversion Operations

-   Convert to List: `list(t)`
-   Convert to Tuple: `tuple([1, 2, 3])`

### Other Operations

-   Unpacking: `a, b = (10, 20)`
-   Delete: `del t`

> **Note:** Tuple does not support adding, removing, or updating
> individual elements.
