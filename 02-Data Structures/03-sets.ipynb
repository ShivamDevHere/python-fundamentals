# Set

``` python
s = {10, 20, 30, 40}
```

### Adding Operations

-   Add: `s.add(50)`
-   Add Multiple: `s.update([60, 70])`

### Removing Operations

-   Remove Value: `s.remove(20)`
-   Safe Remove: `s.discard(20)`
-   Remove Random: `s.pop()`
-   Delete: `del s`
-   Clear: `s.clear()`

### Set Operations

-   Union: `a.union(b)`
-   Intersection: `a.intersection(b)`
-   Difference: `a.difference(b)`
-   Symmetric Difference: `a.symmetric_difference(b)`
-   Subset: `a.issubset(b)`
-   Superset: `a.issuperset(b)`
-   Disjoint: `a.isdisjoint(b)`

### Other Operations

-   Copy: `s.copy()`
-   Check: `20 in s`
-   Length: `len(s)`

------------------------------------------------------------------------
