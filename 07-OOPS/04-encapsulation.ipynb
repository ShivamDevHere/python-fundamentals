### Encapsulation
- it wraps the data (variables) including functions in a single unit. And may restricts some direct access reduces misuse

- dir(Person)  --->  functions publically available
- NGP ---> Not a Good Practice

- Encapsulation using Getter and setter Methods
-  Public, Private, Protected variables or access modifiers
# Public

```python
class Person:
    def __init__(self,name,age):
        self.name = name            #public variable 
        self.age = age

def get_name(self):
    return self.name

def get_age(self):
    return self.age

user1 = Person("Shivam", 23)
get_name(user1)
# dir(user1)    
```

# Private

```python
class Person:
    def __init__(self,name,age):
        self.name  = name
        self.__age = age            #private variable  (dir(Person !=))

def get_name(user1):
    return user1.name

def get_age(user1):
    # return user1.age             # Error [private variable called simply]
    return user1._Person__age      # NGP 

user1 = Person("Shivam", 23)

print(get_name(user1))
try:
    get_age(user1)
except Exception as E:
    print(E)

dir(user1)         # at top '_Person__age' is present as private similarly at bottom public is present
```

# Protected

```python
class Person:
    def __init__(self,name,age):
        self._name = name            #protected variable 
        self.age = age

class Student(Person):
    def __init__(self,name,age):
        super().__init__(name,age)

user1 = Student("Shivam", 23)
print(user1._name,user1.age)
```

##### Using Getter and Setter in Encapsulation

- Getter: gets the value
- Setter: sets the new value

```python
class Bike:
    def __init__(self, name, location):
        self.__name = name
        self.__location = location

    def get_name(self):
        return self.__name
    
    def set_location(self):
        return self.__location
    

Company1 = Bike("TVS", "Kushinagar")
print(Company1.get_name())
```
