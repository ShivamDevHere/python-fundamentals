# Operator Overloading

- Use:Redefining behavior of operators(+, -, *) for custom objects.
- Common operator for Overloading Magic Methods
    - add, sub, mul, truediv, eq, lt, gt
    - `__add__(self, other)`: adds two object using + operator.
    - Similarly for other

```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vetor{self.x, self.y}"

v1 = Vector(3,6)
v2 = Vector(17,21)

print(v1+v2)
```
