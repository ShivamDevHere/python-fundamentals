# Magic Methods

- Definition: = Dunder methods, Py automatically calls them when specific operations are performed on onj.
- Use: We can override, use as we want. 
- Example: Operators Overriding (add, subtract, equal, etc)

#### Some Methods are 

1. `__init__`: Initializes a instance of class.
2. `__str__`: returns string representation of an object.
3. `_repr_`: returns official string representation.
4. `__len__`: returns len(obj)

```python
class Person:
    pass
Developer = Person()
dir(Developer)
```

#### Overriding `__str__`

```python
# Default __str__

class Person:
    pass
    def __init__(self, name, issmart):
        self.name= name
        self.issmart = issmart

Developer = Person("Engineer", True)
print(Developer)
```

```python
# Overridden __str__

class Person:
    pass
    def __init__(self, profession, issmart):
        self.profession= profession
        self.issmart = issmart

    def __str__(self):
        return f"Developer is {self.profession} & issmart is {self.issmart}"

Developer = Person("Engineer", True)
print(Developer)
```

```python
print(repr(Developer))

# Similarly can be Overriden
```
