```python
class Dog:
    # Constructor
    def __init__(self, name, age):
        self.name = name
        self.age = age

# Objects
king_dog = Dog("Dogesh Bhai", "infinity")
print(king_dog)
print(king_dog.name)
print(king_dog.age)
```

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def Meowing(self):
        print(f"A cat \"{self.name}\" of age {self.age} says Meow Meow")

Cat1 = Cat("Mochi",7)
Cat1.Meowing()
```

### Bank account example for class and objects

```python
class Bankaccount:
    def __init__(self,name, balance=0):
        self.name = name
        self.balance = balance
    
    def deposit(self,amount):
        self.balance+=amount
        print(f"Your current balance is: {self.balance}")
    
    def widraw(self,amount):
        if self.balance < amount:
            print("insuffiecient balance")
        else:
            self.balance-=amount
            print(f"Your current balance is: {self.balance}")

    def balance(self):
        print(f"Your current balance: {self.balance}")

yogesh = Bankaccount("Yogesh", 500)
print(yogesh.balance)

yogesh.widraw(100)
yogesh.deposit(100000000)
# for i in range(100000,2000000,100000):
#     print(yogesh.widraw(100000))
```
