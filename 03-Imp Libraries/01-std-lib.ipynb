### Regular Expressions in Python (`re`) EDIT ME

 <!-- The `re` module is used for pattern matching in strings. Commonly used functions are: -->
 <!-- - **`re.search(pattern, text)`** → returns the *first* match of the pattern. -->
 <!-- - **`re.findall(pattern, text)`** → returns *all* non-overlapping matches as a list. -->
 <!-- - **`re.match(pattern, text)`** → checks only at the *beginning* of the string. -->

 <!-- Example: using the pattern `\d+` (one or more digits) on `"There are 67 bananas on 73 trees"`   -->
 <!-- - `re.search` gives **67** (first match).   -->
 <!-- - `re.findall` gives **['67', '73']** (all numbers).   -->


```python
#Regular Expresion
import re

pattern = r'\d+'
text = 'There are bunch of moneys with 67 bananas on 73 trees'
match = re.search(pattern,text)
print(match.group())
```
