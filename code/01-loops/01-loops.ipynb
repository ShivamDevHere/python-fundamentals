# List Comprehension

```python
# Single Line Loop
x = [i for i in range(10)]   # returns List
print(x)
```

```python
# Pair via Loop
even_lst = [num for num in range(10) if num%2==0]

list1 = [1,2,3,4]
list2 = ['abc','bcdef','cgha','d']
list3 = ['abc','bcdef','cgha','d']

pair = [[i,j] for i in list1 for j in list2]

print(pair)

length = [len(word) for word in list3]
```

```python
# Enumerate via Loop
num=[1,2,3,4,5,6,7]

for index,number in enumerate(num):
    print(f"index: {index} value:{number}")
```
