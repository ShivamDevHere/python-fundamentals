# Decorators

A decorator is a function that takes another function (or class) as input, adds some functionality, and returns the modified function without changing its original code.

##### Flow
1. Function copy: copy a func
2. Closure function: func inside func
3. Decorators
---

#### Function copy

```python
def wel():
    return "Hello"

copy = wel
del wel
copy()
```

#### Closure Function

---

```python
def run(msg2="In que"):
    msg = "exiting..."
    def sub_run():
        print(msg2)
        print("processing...")
        print("executing...")
        print(msg)
    return sub_run()
run("started...")
```

```python
def greet(func):
    def sub_greet():
        print("Hello in India is: ", end="")
        func("Namaste")
        print("Bye, Goodluck !")
    return sub_greet()

greet(print)
```

```python
def give_length(func,val):
    
    def calc():
        print(func(val))

    return calc()

give_length(len,"stri")
```

#### Decorators
---

```python
def run(func):
    def sub_run():
        print("processing...")
        print("executing...")
        func()
    return sub_run()

@run
def decorator_func():
    print("did you notice me?")
```
