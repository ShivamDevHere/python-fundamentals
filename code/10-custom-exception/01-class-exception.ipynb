```python
class Error(Exception):
    pass
class DobException(Error):
    pass

year = int(input("Enter year of birth"))
year = 2026 - year

try:
    if year > 20 and year <30:
        print("Allowed")
    else: 
        raise DobException
except DobException:
    print(f"Your age: {year} is not between 20 & 30")
```
