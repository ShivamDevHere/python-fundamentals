# Abstraction

- Defination: Hiding complex implementation and only showing necessary features of object.

- Example: Washing machine abstracting circuits by buttons and display

```python
from abc import ABC, abstractmethod

class Machine(ABC):
    def operate(self):
        print("Operating...")
    def off(self):
        print("Power Cut-off")

    @abstractmethod
    def start(self):
        print("Power received")
        # pass

class Lamp(Machine):
    def start(self):
        print("Lamp is light")



blue1 = Lamp()
blue1.start()
blue1.off()
```
