# Magic Methods

- Definition: = Dunder methods, Py automatically calls them when specific operations are performed on onj.
- Use: We can override, use as we want.

#### Some Methods are 

1. `__init__`: Initializes a instance of class.
2. `__str__`: returns string representation of an object.
3. `_repr_`: returns official string representation.
4. `__len__`: returns len(obj)
