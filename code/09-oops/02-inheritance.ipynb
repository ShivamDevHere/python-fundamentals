## Inheritance
##### Inheriting properties from parents for child (unidirectional); prop. = atr obj fun()

```python
# Parent Class

class Car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors   = doors
        self.enginetype = enginetype
    
    def drive(self):
        print(f"Yes {Models_car[0]} can drive via a {self.enginetype} Engine")
        return "ALL Operation Completed from Car Class"

Models_car = ["Oddi", "BMW"]

Oddi = Car(6,4,"IC")

print(Oddi.drive())
```

##### Single Inheritance

```python
class Hybrid(Car):
    def __init__(self,windows,doors,enginetype,selfdrive,name):
        super().__init__(windows,doors,enginetype)
        self.selfdrive = selfdrive
        self.name = name
    
    def automatic(self):
        print(f"Yes, {self.name} is automatic with {self.enginetype} Engine")


Tesla = Hybrid(8,4,"Motor",True,"Tesla").automatic()
    

    
```

##### Multiple Inheritance

```python
# Base Class 1

class Animal:
    def __init__(self,owner):
        self.owner = owner
    
    def speaks(self):
        print("I speak hehe, sometimes")

# Base Class 2

class Pet:
    def __init__(self,name):
        self.name = name

# Inheriting from Class 1 2

class Behaviour(Animal,Pet):
    def __init__(self,owner,name,sound):
        Animal.__init__(self,owner)
        Pet.__init__(self,name)
        self.sound = sound

    def speaks(self):
        return f"So, {self.name} of {self.owner} says {self.sound}"   

Dog = Behaviour("Krish", "Dog", "Woof Woof").speaks()  
print(Dog)
```
