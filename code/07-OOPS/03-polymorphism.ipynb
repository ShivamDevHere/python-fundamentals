### Polymorphism
##### obj of diff classes is treated as objects of SUPER class; Provides ways to perform same name of operation with diff ways: [Method overriding]


```python
# Base Class
class Animal:
    def speak(self):
        return "It is animal sound"
    
# Derived Class 1
class Cat(Animal):
    def speak(self):
        return "Meow Meow"

# Derived Class 2
class Dog(Animal):
    def speak(self):
        return "Woof Woof!"

Chuimui = Cat()
kutta = Animal()
kutta.speak()
Chuimui.speak()
kutta.speak()
```

#### Another Example for Polymorphism with functions

```python
class Shape:
    def area(self):
        print("this is a area, is it?")

class Rectangle(Shape):
    def __init__(self,width,height):
        self.width = width
        self.height = height

    def area(self):
        return f"area of rec is {self.width*self.height}"
    
class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius

    def area(self):
        return f"Area for cir is :{3.14* self.radius**2}"
    
circle = Circle(3)
rectangle = Rectangle(2,3)

print(rectangle.area())
print(circle.area())
```

### RT Instance : Abstract Base Class

```python
from abc import ABC,abstractmethod

# Defined an abstract class
class Vehicle(ABC):
    @abstractmethod
    def start_engine(self):
        pass

# Derived class 1
class Car(Vehicle):
    def start_engine(self):
        return "Car engine started"
    
# Derived class 2
class Motorcycle(Vehicle):
    def start_engine(self):
        return "Motorcycle started"
    
# Function that demonstrate polymorphism
def start_vehicle(vehicle):
    print(vehicle.start_engine())


# Objects created for class

car = Car()
bike = Motorcycle()

start_vehicle(car)
start_vehicle(bike)
```
