# Generators

- way to create iterators (series of lazy values)
- keyword: `yield`
- generates, do not store

```python
def square(val):
    for val in range(val):
        yield val**2    

square(2)
```

```python
for i in square(7):
    print(i)
```

```python
a= square(5)
print(a)
```

```python
next(a)
```

```python
# returns multiple values

def new_generator(val):
    yield 1
    yield 2
    yield 3

for i in new_generator(66):
    print(i)
```

# Practical Use Case

```python
def read_large_file(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line
```

```python
file_path = "large-file.txt"

for line in read_large_file(file_path):
    print(line.strip())
```
