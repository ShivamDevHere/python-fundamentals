```python
x = int(input(""))


def primeproduct(m):
    Flag = True
    Lprime=[]
    if m<=1:
        return False
    else:
        for i in range(2,m):
            if m%i==0:
                Lprime.append(i)
    z=Lprime
    for i in z:
        for j in range(2,i):
            if i%j==0:
                Flag = False
                break
            
    return Flag

print(primeproduct(x))
```
